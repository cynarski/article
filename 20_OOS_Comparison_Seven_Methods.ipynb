# Seven-method OOS warning timelines

Reproduce publication figures comparing LPPLS-CI, BSADF, Logistic Regression, Decision Tree, XGBoost, LightGBM and CatBoost.

The two statistical rules and the five chronologically tuned ML models have different calibration protocols, but all share the same OOS dates and the same future 10%/60-day drop target. The notebook checks both conditions before plotting.


```python
from pathlib import Path
import subprocess, sys

ROOT = Path('.').resolve()
PRED_TUNED = ROOT / '08b_wyniki_tuning' / '08b_04_predykcje_OOS.csv'
PRED_FIXED = ROOT / '18_wyniki_bsadf_future_drop' / 'drop_10_h60' / 'OOS_predictions.csv'
REFERENCE = ROOT / '13_wyniki_bubble' / '13_bubble_reference_daily.csv'
OUT = ROOT / '20_wyniki_multi_method_timelines'
for file in (PRED_TUNED, PRED_FIXED, REFERENCE):
    if not file.is_file():
        raise FileNotFoundError(f'Missing OOS input: {file}')
subprocess.run([sys.executable, str(ROOT / 'scripts' / 'plot_oos_multi_method_timelines.py'),
                '--pred-08b', str(PRED_TUNED), '--pred-18', str(PRED_FIXED),
                '--reference', str(REFERENCE), '--out', str(OUT)], check=True)
```

```python
import pandas as pd
from IPython.display import display
summary = pd.read_csv(OUT / 'oos_seven_method_metadata.csv')
display(summary)
print('Vector figures:', OUT / 'oos_seven_methods_compact.pdf',
      OUT / 'oos_seven_methods_price_panels.pdf', sep='\n')
```
