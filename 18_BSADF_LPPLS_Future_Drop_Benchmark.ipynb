# 18 — Sampled-BSADF vs LPPLS vs ML: future BTC corrections

Compare statistical and machine-learning risk rankings against a **future price decline >=10% within 60 days** (the independently constructed 07/08b label). This is an early-warning benchmark and **does not equate every decline with a bursting economic bubble**.

The statistical BSADF comparator is the **last available test statistic minus its simulated critical value**. LPPLS CI uses all raw-window fits in its denominator. ML variants use price, LPPLS and/or BSADF features on identical OOS dates. Nested tuning in notebook 08b is distinct; this benchmark fixes XGBoost hyperparameters to isolate feature-family differences.
