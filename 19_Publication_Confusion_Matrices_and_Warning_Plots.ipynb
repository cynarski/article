# 19 — Publication figures from chronological OOS forecasts

This notebook regenerates English-language paper figures without fitting models or tuning thresholds on test labels. It combines (1) tuned 08b predictions for a >=15% future decline at 30/45/60 days and (2) fixed-configuration 18 predictions for a >=10% future decline at 60 days. These are different experiments and must not be presented as one ranking.

Required preceding notebooks: **08b**, **13**, **18**. Place their output CSVs under the standard output directories.

```python
from pathlib import Path
import subprocess, sys
ROOT=Path('.').resolve()
PRED08=ROOT/'08b_wyniki_tuning'/'08b_04_predykcje_OOS.csv'
PRED18=ROOT/'18_wyniki_bsadf_future_drop'/'drop_10_h60'/'OOS_predictions.csv'
REF=ROOT/'13_wyniki_bubble'/'13_bubble_reference_daily.csv'
OUT=ROOT/'19_wyniki_publication_figures'
for path in (PRED08,PRED18,REF):
    if not path.is_file(): raise FileNotFoundError(f'Missing required OOS input: {path}')
subprocess.run([sys.executable,str(ROOT/'scripts'/'plot_oos_publication_figures.py'),
                '--pred-08b',str(PRED08),'--pred-18',str(PRED18),
                '--reference',str(REF),'--output',str(OUT)],check=True)
```

## Interpretation safeguards

The 15% matrices use the already tuned 08b walk-forward XGBoost model; the 10% event comparison uses the fixed-configured 18 benchmark. UNION shading is retrospective context, not a future training label. Drawdown-event onsets use a shifted 180-day peak with a 60-day minimum event separation; unconfirmed warning starts are evaluated against the **different** 10%/60-day future-decline label. The first seven test days are onset-censored. Thresholds are not reselected on OOS.
